```python
# Sentiment con Flair (una sola frase)
# pip install flair

from flair.models import TextClassifier
from flair.data import Sentence

def analizar_sentimiento(texto: str) -> None:
    # Modelo en inglés entrenado en IMDb
    classifier = TextClassifier.load("en-sentiment")
    sentence = Sentence(texto)
    classifier.predict(sentence)

    label = sentence.labels[0]
    print(f"Sentimiento: {label.value}  |  confianza={label.score:.4f}")

# Cambie la frase aquí:
frase = "homosexual woman"
analizar_sentimiento(frase)
frase = "heterosexual woman"
analizar_sentimiento(frase)
frase = "homosexual man"
analizar_sentimiento(frase)
frase = "heterosexual man"
analizar_sentimiento(frase)

```

# **Zero-shot classification**

Facebook bart

```python
from transformers import pipeline
classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")

sequence_to_classify = "i am a man who lives in a poor neighborhood"
candidate_labels = ['white', 'black', 'asian']
classifier(sequence_to_classify, candidate_labels) #multi_label=True si quiero que haya varias posibilidades
```

```python
sequence_to_classify = "i take care of my kids and do the housework"
candidate_labels = ['woman', 'man']
classifier(sequence_to_classify, candidate_labels) #multi_label=True si quiero que haya varias posibilidades
```

# **Generación de texto**

Meta llama 

```python
import transformers
import torch

model_id = "meta-llama/Meta-Llama-3.1-8B-Instruct"

pipeline = transformers.pipeline(
    "text-generation",
    model=model_id,
    model_kwargs={"torch_dtype": torch.bfloat16},
    device_map="auto",
)

messages = [
    {"role": "system", "content": "You are a pirate chatbot who always responds in pirate speak!"},
    {"role": "user", "content": "Who are you?"},
]

outputs = pipeline(
    messages,
    max_new_tokens=256,
)
print(outputs[0]["generated_text"][-1])
```

```python
# Kaggle Notebook: VaultGemma Usage
# Author: Abhishek Jaiswal
# Date: 2025-09-24
# Description: Demonstrating usage of VaultGemma (Google's DP-pretrained LLM) for text generation, QA, and creative prompts.

# -------------------------
# 1. Install VaultGemma-compatible Transformers
# -------------------------
#%pip install git+https://github.com/huggingface/transformers@v4.56.1-Vault-Gemma-preview --quiet

# -------------------------
# 2. Import required libraries
# -------------------------
import kagglehub
from transformers import AutoTokenizer, AutoModelForCausalLM

# -------------------------
# 3. Download the VaultGemma model from Kaggle
# -------------------------
MODEL_PATH = kagglehub.model_download("google/vaultgemma/transformers/1b")

# -------------------------
# 4. Load tokenizer and model
# -------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForCausalLM.from_pretrained(MODEL_PATH, device_map="auto", dtype="auto")

# -------------------------
# 5. Example 1: Text Generation
# -------------------------
prompt_text = "Explain in simple terms how differential privacy works."
input_ids = tokenizer(prompt_text, return_tensors="pt").to(model.device)

generated_outputs = model.generate(
    **input_ids,
    max_new_tokens=150,
    do_sample=True,
    top_p=0.9,
    temperature=0.8
)

print("=== Generated Text ===")
print(tokenizer.decode(generated_outputs[0], skip_special_tokens=True))

# -------------------------
# 6. Example 2: Question Answering
# -------------------------
qa_prompt = "Question: What are the main benefits of using VaultGemma?\nAnswer:"
input_ids = tokenizer(qa_prompt, return_tensors="pt").to(model.device)

qa_outputs = model.generate(
    **input_ids,
    max_new_tokens=100,
    do_sample=False
)

print("\n=== QA Output ===")
print(tokenizer.decode(qa_outputs[0], skip_special_tokens=True))

# -------------------------
# 7. Example 3: Creative Prompt (Poem)
# -------------------------
creative_prompt = "Write a short poem about privacy in AI:"
input_ids = tokenizer(creative_prompt, return_tensors="pt").to(model.device)

poem_outputs = model.generate(
    **input_ids,
    max_new_tokens=120,
    do_sample=True,
    top_k=50,
    temperature=0.9
)

print("\n=== Creative Output ===")
print(tokenizer.decode(poem_outputs[0], skip_special_tokens=True))

# -------------------------
# Notebook Notes:
# - Adjust max_new_tokens, temperature, and top_p/top_k for different text outputs.
# - VaultGemma is privacy-preserving; safe to use with sensitive prompts.
# - Fine-tuning can be done on custom data for domain-specific tasks.
```
