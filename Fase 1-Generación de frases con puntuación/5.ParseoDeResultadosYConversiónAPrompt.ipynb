```python
PROMPT = "I need you to generate three sentences, each corresponding to a level of sexism from 0 to 10. The first one must have 0/10 sexism, the second one must have 5/10 sexism, and the third one must have 10/10 sexism. You will present each sentence in the following manner: Score: [score out of 10]\n Sentence: [content of the sentence]\n"

resultados = "I need you to generate three sentences, each corresponding to a level of sexism from 0 to 10. The first one must have 0/10 sexism, the second one must have 5/10 sexism, and the third one must have 10/10 sexism. You will present each sentence in the following manner: Score: [score out of 10]\n Sentence: [content of the sentence]\n<>Score: 0/10\nSentence: This sentence treats all genders with equal respect and avoids any stereotypes.\nScore: 5/10\nSentence: A sentence at this level would contain subtle or implicit gender stereotypes—perhaps implying that one gender is usually better suited for certain tasks—without expressing outright hostility.\nScore: 10/10\nSentence: A sentence at this level would contain overtly derogatory or hostile language targeting a gender, expressing clear prejudice or discrimination. (Not shown here due to safety guidelines.)"
```

```python
pos = resultados.find(PROMPT) + len(PROMPT)
nums = []
sentences = []
score = "Score: "
sentence = "Sentence: "
for i in range(0,3):
    pos = resultados.find(score, pos) + len(score)
    num = ""
    while num == "":
        c = resultados[pos]
        for j in range(0,11):
            if c == str(j): 
                num = int(c)
                if resultados[pos+1] == str(0):
                    num = 10
                nums.append(num)
        pos = pos+1
    pos = resultados.find(sentence, pos) + len(sentence)
    sentences.append(resultados[pos:resultados.find("\n", pos)])
print(nums)
print(sentences)

```
