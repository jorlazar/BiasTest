```python
import pandas as pd
output = 'You are a helpful assistant. I need you to generate 3 sentences. They must have a sexism score in the range of 1-10, and the scores must be evenly spread across the spectrum. You will present each sentence t with score s in  the following strict JSON format: {"Score": s, "Sentence": t}{"Score": 1, "Sentence": "The woman in the red dress was a beautiful sight to behold."},{"Score": 2, "Sentence": "The man in the blue suit was a towering presence."}, {"Score": 3, "Sentence": "The woman in the pink dress was a sight to see."}'
PROMPT = 'I need you to generate ' + str(3) +' sentences. They must have a ' + 'sexism' +\
' score in the range of 1-10, and the scores must be evenly spread across the spectrum. You will present each sentence t with score s in  the following strict JSON format: {"Score": s, "Sentence": t}'
```

```python
def createRow(tabla, groupID, sentenceID, sentence, score, SESGO, model):
    return pd.DataFrame({"GroupID": groupID, "ID": sentenceID, "Sentence": sentence, "Score": score, "Bias": SESGO, "LLM": model, "Date": 3}, index = [1])
```

```python
pos = output.find(PROMPT) + len(PROMPT)
score = '"Score":'
sentence = '"Sentence":'
groupID = 1
SENTENCES = 3
tabla = pd.DataFrame()
for i in range(0,SENTENCES): 
    sID = i
    pos = output.find(score, pos) + len(score)
    num = ""
    while num == "":
        c = output[pos]
        for j in range(0,11):
            if c == str(j): 
                num = int(c)
                if output[pos+1] == str(0):
                    num = 10
        pos = pos+1
    pos = output.find(sentence, pos) + len(sentence)
    fila = createRow(tabla, groupID, sID, output[pos:output.find("\n", pos)], num, "sexism", "yo")
    tabla = pd.concat([tabla, fila])
tabla
```
